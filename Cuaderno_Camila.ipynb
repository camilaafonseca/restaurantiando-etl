```python
import pandas as pd

df_data1 = pd.read_excel('/content/dataset_tripadvisor_2026-05-18_16-00-01-293.xlsx', sheet_name='Data1')

# Convertir a minúsculas solo para la búsqueda interna
df_strings = df_data1.astype(str).apply(lambda x: x.str.lower())

# =========================================================
# CREACIÓN DE BOOLEANOS: MOMENTOS DE CONSUMO (TIPO DE COMIDA)
# =========================================================

# 1. Desayunando
df_data1['Desayunando'] = df_strings.apply(lambda row: row.str.contains('desayuno|breakfast').any(), axis=1)

# 2. Brunchando (Brunch)
df_data1['Brunchando'] = df_strings.apply(lambda row: row.str.contains('brunch').any(), axis=1)

# 3. Almorzando (Almuerzo)
df_data1['Almorzando'] = df_strings.apply(lambda row: row.str.contains('almuerzo|lunch').any(), axis=1)

# 4. Picando
df_data1['Picando'] = df_strings.apply(lambda row: row.str.contains('picadas|tapas|para picar|entradas').any(), axis=1)

# 5. Tardeando
df_data1['Tardeando'] = df_strings.apply(lambda row: row.str.contains('café|repostería|onces|tarde|afternoon').any(), axis=1)

# 6. Cocteleando
df_data1['Cocteleando'] = df_strings.apply(lambda row: row.str.contains('cocteles|cócteles|bar|pub|cocteleando').any(), axis=1)

# 7. Tomando algo / Bebidas (Más amplio)
df_data1['Tomando algo'] = df_strings.apply(lambda row: row.str.contains('bebidas|drinks|bebidas alcohólicas|vino|cerveza').any(), axis=1)

# 8. Cenando (Cena)
df_data1['Cenando'] = df_strings.apply(lambda row: row.str.contains('cena|dinner').any(), axis=1)

# 9. Trasnochando / Abierto hasta tarde
df_data1['Trasnochando'] = df_strings.apply(lambda row: row.str.contains('abierto hasta tarde|trasnochar|late night|madrugada').any(), axis=1)


# =========================================================
# SELECCIÓN E IDENTIFICADOR PARA EL JOIN
# =========================================================

# ¡CRUCIAL! Asegúrate de incluir la columna del identificador único.
# En Apify para TripAdvisor suele llamarse 'id', 'id_restaurante', 'placeId' o 'url'.
# Reemplaza 'id' por el nombre exacto que tenga la columna de tu ID en tu Excel.
columna_id = 'id'

nuevas_columnas_momentos = [
    'Desayunando', 'Brunchando', 'Almorzando', 'Picando', 'Tardeando',
    'Cocteleando', 'Tomando algo', 'Cenando', 'Trasnochando'
]

# Creamos el DataFrame final con el ID y las nuevas clasificaciones
df_momentos_final = df_data1[[columna_id] + nuevas_columnas_momentos]

# Guardar para el JOIN
df_momentos_final.to_excel("tabla_momentos_para_join.xlsx", index=False)

print("¡Listo! Columnas de momentos de consumo creadas e ID conservado para el JOIN.")
```

```python
import pandas as pd
import numpy as np

# 1. Cargar datos
df_data1 = pd.read_excel('/content/dataset_tripadvisor_2026-05-18_16-00-01-293.xlsx', sheet_name='Data1')

# Convertimos todo el dataframe a minúsculas en un objeto temporal para buscar fácil
df_strings = df_data1.astype(str).apply(lambda x: x.str.lower())

# =========================================================
# LÓGICA 1: CONSOLIDAR EL AMBIENTE / ESTILO
# =========================================================
def determinar_ambiente(fila):
    texto_completo = " ".join(fila) # Une todo el texto de las columnas de esa fila

    if any(palabra in texto_completo for palabra in ['gourmet', 'alta cocina', 'fine dining']):
        return 'Gourmet / Alta cocina'
    elif any(palabra in texto_completo for palabra in ['fast food', 'comida rápida', 'food truck']):
        return 'Comida rápida / Fast food'
    elif 'fast casual' in texto_completo:
        return 'Fast casual'
    elif 'buffet' in texto_completo:
        return 'Buffet'
    elif 'temático' in texto_completo or 'tematico' in texto_completo:
        return 'Temático'
    elif 'familiar' in texto_completo:
        return 'Restaurante familiar'
    elif any(palabra in texto_completo for palabra in ['cafetería', 'cafeteria', 'bistró', 'bistro', 'café', 'cafe']):
        return 'Cafetería / Bistró'
    elif any(palabra in texto_completo for palabra in ['casual', 'casual dining']):
        return 'Casual'

    return 'No especificado' # Si no encuentra ninguna coincidencia

# =========================================================
# LÓGICA 2: CONSOLIDAR LA ESPECIALIDAD / TIPO DE COCINA
# =========================================================
def determinar_especialidad(fila):
    texto_completo = " ".join(fila)

    if any(palabra in texto_completo for palabra in ['mariscos', 'pescado', 'mediterránea', 'mediterranea']):
        return 'Comida de mar'
    elif any(palabra in texto_completo for palabra in ['italiana', 'pizzería', 'pizzeria', 'pizza']):
        return 'Italiana / Pizzería'
    elif any(palabra in texto_completo for palabra in ['parrilla', 'parrillada', 'churrasquería', 'churrasqueria', 'carnes', 'asado']):
        return 'Parrilla / Carnes'
    elif any(palabra in texto_completo for palabra in ['colombiana', 'criolla', 'local']):
        return 'Colombiana'
    elif any(palabra in texto_completo for palabra in ['mexicana', 'mexican']):
        return 'Mexicana'
    elif any(palabra in texto_completo for palabra in ['japonesa', 'japanese', 'sushi', 'asiática', 'asiatica']):
        return 'Asiática / Sushi'
    elif any(palabra in texto_completo for palabra in ['libanesa', 'medio oriente', 'árabe', 'arabe']):
        return 'Medio Oriente'
    elif any(palabra in texto_completo for palabra in ['francesa', 'french']):
        return 'Francesa'
    elif any(palabra in texto_completo for palabra in ['bar', 'cervecería', 'pub']):
        return 'Bar / Pub'
    elif 'fusión' in texto_completo or 'fusion' in texto_completo:
        return 'Fusión'
    elif any(palabra in texto_completo for palabra in ['internacional', 'étnico', 'etnico']):
        return 'Internacional'

    return 'Otras cocinas'

# =========================================================
# APLICAR LAS FUNCIONES Y GENERAR EL ARCHIVO DEFINITIVO
# =========================================================

# Creamos un nuevo DataFrame para no dañar el original
df_resultado = pd.DataFrame()

# Identifica la columna de tu ID (reemplaza 'id' si se llama de otra forma, ej: 'placeId' o 'name')
columna_id = 'id'
df_resultado[columna_id] = df_data1[columna_id]

# Aplicamos la lógica fila por fila (axis=1) buscando en el df en minúsculas
df_resultado['Ambiente'] = df_strings.apply(determinar_ambiente, axis=1)
df_resultado['Especialidad'] = df_strings.apply(determinar_especialidad, axis=1)

# Guardar el resultado en un Excel super limpio
df_resultado.to_excel("/content/dataset_clasificado_limpio.xlsx", index=False)

print("¡Proceso completado con éxito! Revisa tu archivo 'dataset_clasificado_limpio.xlsx'")
```
